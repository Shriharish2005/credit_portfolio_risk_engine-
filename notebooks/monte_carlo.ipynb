```python
import os, sys
sys.path.insert(0, os.path.abspath('..'))
import numpy as np
import pandas as pd
from backend.risk_engine.monte_carlo import MonteCarloSimulationEngine

loans_df = pd.read_csv('../data/synthetic/loans.csv')
sample = loans_df.sample(n=2000, random_state=42)
engine = MonteCarloSimulationEngine(random_seed=42)
res = engine.run_simulation(sample['pd'].values, sample['lgd'].values, sample['ead'].values, n_simulations=5000, asset_correlation=0.18)
print(f'Mean Loss: ${res["mean_loss"]:,.2f}')
print(f'95% VaR:   ${res["var_95"]:,.2f}')
print(f'99% VaR:   ${res["var_99"]:,.2f}')
print(f'99% ES:    ${res["es_99"]:,.2f}')
pd.Series(res['percentiles'])
```
